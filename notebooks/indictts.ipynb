```python
from typing import Literal
import torch
from TTS.api import TTS
from pprint import pprint as print

# Set device
device: Literal['cuda','cpu'] = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# Initialize TTS model
tts = TTS(
    model_path="/home/ninurai/projects/tts/coqui-fastapi-streaming/server/notebooks/models/v1/as/fastpitch/best_model.pth",
    config_path="/home/ninurai/projects/tts/coqui-fastapi-streaming/server/notebooks/models/v1/as/fastpitch/config.json",
    vocoder_path="/home/ninurai/projects/tts/coqui-fastapi-streaming/server/notebooks/models/v1/as/hifigan/best_model.pth",
    vocoder_config_path="/home/ninurai/projects/tts/coqui-fastapi-streaming/server/notebooks/models/v1/as/hifigan/config.json",
    gpu=True
)

# Ensure all model components are on GPU
if hasattr(tts, 'synthesizer'):
    # Move TTS model
    tts.synthesizer.tts_model.to(device)
    # Move all embeddings to GPU
    if hasattr(tts.synthesizer.tts_model, 'emb_g'):
        tts.synthesizer.tts_model.emb_g.to(device)
    # Move vocoder
    if hasattr(tts.synthesizer, 'vocoder'):
        tts.synthesizer.vocoder.to(device)
    # Set use_cuda flag
    tts.synthesizer.use_cuda = True

# Set model to eval mode
tts.synthesizer.tts_model.eval()
if hasattr(tts.synthesizer, 'vocoder'):
    tts.synthesizer.vocoder.eval()

# Now try the TTS generation
with torch.inference_mode():
    wav = tts.tts(
        text="""হেৰুৱা স্মৃতিৰ গান

কুঁৱলীৰ দৰে মন, হেৰাই যোৱা সপোন,
নিৰৱতাৰ গান,  বুকুৰ শূন্যস্থান।
শেৱালিৰ সুবাস,  উৰি গ'ল দূৰলৈ,
একাডেমীয়া ৰাতি,  চন্দ্ৰৰ নীৰৱতাই।

নদীৰ সোঁতত,  ভাসি গ'ল সময়,
হেৰুৱা প্ৰেমৰ,  অশ্ৰুৰ  নদী বয়।
স্মৃতিৰ  গভীৰত,  জীৱনৰ  গান,
হেৰুৱা  সময়ৰ,  বিলাপৰ  তান।

শূন্য  হাত,  শূন্য  মন,  শূন্য  আশা,
হেৰুৱা  প্ৰেমৰ,  এয়াই  ভাষা।
তথাপিও  জীৱন,  চলিয়েই  থাকে,
নতুন  সপোনে,  আকৌ  মাতে।""",
        speaker="male",
        file_path="output.wav"
    )
```
